```python
import requests
import zipfile
import io

url = "https://naciscdn.org/naturalearth/110m/cultural/ne_110m_admin_0_countries.zip"
response = requests.get(url)
with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    z.extractall("temp_shp")
world = gpd.read_file("temp_shp/ne_110m_admin_0_countries.shp")

# Merge
world_data = world.merge(df_grouped, left_on='NAME', right_on='pais_rnaturalearth', how='left')
```

```python

# Plot choropleth
fig, ax = plt.subplots(1, 1, figsize=(15, 10))
world_data.plot(column='percentual', ax=ax, legend=True, cmap='OrRd', missing_kwds={'color': 'lightgrey'})
ax.set_title('Percentage of Emergency Hospitalizations by Country')
plt.savefig('mapa_percentual_en.png', dpi=300, bbox_inches='tight')
plt.close()

# Portuguese version of map
fig, ax = plt.subplots(1, 1, figsize=(15, 10))
world_data.plot(column='percentual', ax=ax, legend=True, cmap='OrRd', missing_kwds={'color': 'lightgrey'})
ax.set_title('Percentual de Internações de Urgência por País')
plt.savefig('mapa_percentual_pt.png', dpi=300, bbox_inches='tight')
plt.close()

# Now for the time series by continent
```

```python



# Now, group by ANO_CMPT and continente

# Pivot
df_pivot = estrangeiros_por_continente_ano.pivot(index='ANO_CMPT', columns='continentes', values='qtd').fillna(0)

# English version
df_pivot_en = df_pivot.copy()
# Sort columns by total sum ascending (smaller areas at bottom)
df_pivot_en = df_pivot_en[df_pivot_en.sum().sort_values().index]
```

```python



# Plot stacked area English
ax = df_pivot_en.plot.area(figsize=(10, 6), stacked=True)

def mil_formatter_en(x, pos):
    if x >= 1000:
        return f"{int(x/1000)}k"
    else:
        return str(int(x))

plt.gca().yaxis.set_major_formatter(FuncFormatter(mil_formatter_en))

plt.title("Annual emergency hospital admissions of foreign nationals in Brazil's Unified Health System (SUS) by continent")
plt.xlabel("Year")
plt.ylabel("Number of hospitalizations")
plt.legend(title='Continents', bbox_to_anchor=(1.05, 1), loc='upper left')

# Add labels
max_y = df_pivot_en.sum(axis=1).max()
for i, row in df_pivot_en.iterrows():
    valor = row.sum()
    label = mil_formatter_en(valor, None)
    plt.text(i, valor + max_y * 0.02, label, ha="center", va="bottom", fontsize=10)

plt.figtext(0.5, -0.01, "Source: SIH/SUS – Sistema de Informações Hospitalares do SUS (Datasus).", ha="center", fontsize=9, style="italic")
plt.figtext(0.5, -0.05, "Note: 2025 data is partial (up to August 2025)", ha="center", fontsize=9, style="italic")

plt.subplots_adjust(bottom=0.2)
plt.tight_layout()
plt.savefig('grafico_area_en.png', dpi=300, bbox_inches='tight')
plt.close()

# Portuguese version
# Translate continents to Portuguese
continent_translation = {
    'Africa': 'África',
    'Asia': 'Ásia',
    'Europe': 'Europa',
    'North America': 'América do Norte',
    'South America': 'América do Sul',
    'Oceania': 'Oceania',
    'Antarctica': 'Antártica'
}
df_pivot_pt = df_pivot.rename(columns=continent_translation)
# Sort columns by total sum ascending (smaller areas at bottom)
df_pivot_pt = df_pivot_pt[df_pivot_pt.sum().sort_values().index]

# Plot stacked area Portuguese
ax = df_pivot_pt.plot.area(figsize=(10, 6), stacked=True)

def mil_formatter_pt(x, pos):
    if x >= 1000:
        return f"{int(x/1000)} mil"
    else:
        return str(int(x))

plt.gca().yaxis.set_major_formatter(FuncFormatter(mil_formatter_pt))

plt.title("Internações com urgência de pessoas estrangeiras no SUS por continente")
plt.xlabel("Ano")
plt.ylabel("Quantidade de internações")
plt.legend(title='Continentes', bbox_to_anchor=(1.05, 1), loc='upper left')

# Add labels
max_y = df_pivot_pt.sum(axis=1).max()
for i, row in df_pivot_pt.iterrows():
    valor = row.sum()
    label = mil_formatter_pt(valor, None)
    plt.text(i, valor + max_y * 0.02, label, ha="center", va="bottom", fontsize=10)

plt.figtext(0.5, -0.01, "Fonte: SIH/SUS – Sistema de Informações Hospitalares do SUS (Datasus).", ha="center", fontsize=9, style="italic")
plt.figtext(0.5, -0.05, "Nota: Dados de 2025 são parciais (até agosto de 2025)", ha="center", fontsize=9, style="italic")

plt.subplots_adjust(bottom=0.2)
plt.tight_layout()
plt.savefig('grafico_area_pt.png', dpi=300, bbox_inches='tight')
plt.close()
```
