```python
import pandas as pd
import os
import warnings
warnings.filterwarnings("ignore")
```

```python
folder_path = 'saida_sih'

all_files = [f for f in os.listdir(folder_path) if f.startswith('sih_') and f.endswith('.csv')]

file_paths = [os.path.join(folder_path, f) for f in all_files]

dfs = []

for file in file_paths:
    print(f'Lendo arquivo {file}...')
    df_tmp = pd.read_csv(file, dtype={'MUNIC_RES': str, 'SEXO': str, 'UTI_MES_TO': int,
                                      'DT_INTER': str, 'DIAG_PRINC': str, 'IDADE': int,
                                      'CAR_INT': str, 'UF': str, 'ANO': int})
    dfs.append(df_tmp)
```

```python
df_full = pd.concat(dfs, ignore_index=True)
print(f'Todos os arquivos lidos e concatenados. Total de linhas: {df_full.shape[0]}')
```

```python
df_full.to_csv("dados_sih_completos.csv")
```
