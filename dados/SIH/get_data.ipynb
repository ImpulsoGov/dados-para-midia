```python
from pysus import SIH
import pandas as pd
import os
sia = SIH().load()
```

```python
variaveis = ["MUNIC_RES", "SEXO", "UTI_MES_TO", "DT_INTER", "DIAG_PRINC","IDADE", "CAR_INT"]
```

```python
sih = SIH().load()

ufs = [
    'AC', 'AL', 'AP', 'AM', 'BA', 'CE', 'DF', 'ES', 'GO',
    'MA', 'MT', 'MS', 'MG', 'PA', 'PB', 'PR', 'PE', 'PI',
    'RJ', 'RN', 'RS', 'RO', 'RR', 'SC', 'SP', 'SE', 'TO'
]

ufs = ['SE', 'TO']
anos = [2022, 2023, 2024]

os.makedirs("saida_sih", exist_ok=True)

for uf in ufs:
    dados_uf = []

    for ano in anos:
        try:
            arquivos = sih.get_files(group="RD", year=ano, uf=uf)
            if not arquivos:
                print(f"Nenhum arquivo encontrado para {uf}/{ano}")
                continue

            for arq in arquivos:
                parquet = sih.download(arq)
                df = parquet.to_dataframe()
                df = df[variaveis]
                df = df[df['DIAG_PRINC'].str.startswith('I', na=False)]
                df["UF"] = uf
                df["ANO"] = ano
                dados_uf.append(df)
                print(f"Adicionado {uf}/{ano}")

        except Exception as e:
            print(f"Erro ao processar {uf}/{ano}: {e}")

    if dados_uf:
        df_uf = pd.concat(dados_uf, ignore_index=True)
        df_uf.to_csv(f"saida_sih/sih_{uf}_2022_2024_.csv", index=False)
        print(f"Salvo: sih_{uf}_2022_2024.csv")
```
