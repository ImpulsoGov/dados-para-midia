```python
import pandas as pd
import os
import glob
import warnings
warnings.filterwarnings("ignore")
import re
```

### Concatenar dataframes

```python
caminho_pasta = "./" 
arquivos = glob.glob(os.path.join(caminho_pasta, "*.csv"))

dfs = []

for arquivo in arquivos:
    df = pd.read_csv(arquivo, sep=";", dtype={'DTNASC': str}, encoding="ISO-8859-1")
    df["arquivo_origem"] = os.path.basename(arquivo)  
    dfs.append(df)

df = pd.concat(dfs, ignore_index=True)

print(df.head())
#df.to_csv("dados_completos_mortalidade.csv")
```

```python
df['DTOBITO'] = df['DTOBITO'].astype(str).str.zfill(8)
df['DTOBITO'] = pd.to_datetime(df['DTOBITO'].astype(str), format='%d%m%Y')
df['DTNASC'] = df['DTNASC'].astype(str).str.zfill(8)
df['DTNASC'] = pd.to_datetime(df['DTNASC'].astype(str), format='%d%m%Y', errors='coerce')
df['ano_obito'] = df['DTOBITO'].dt.year
```

```python
df['CID3'] = df['CAUSABAS'].str[:3]
```

```python
df = df[
['ORIGEM', 'TIPOBITO', 'DTOBITO', 'HORAOBITO', 'NATURAL',
       'CODMUNNATU', 'DTNASC', 'IDADE', 'SEXO', 'RACACOR', 'ESTCIV', 'ESC',
       'ESC2010', 'OCUP', 'CODMUNRES', 'LOCOCOR', 
       'CODMUNOCOR','GRAVIDEZ', 'SEMAGESTAC',
       'GESTACAO', 'PARTO', 'OBITOPARTO', 'OBITOGRAV',
       'OBITOPUERP', 'ASSISTMED', 'EXAME', 'CIRURGIA', 'CAUSABAS', 'CB_PRE','CAUSAMAT', 'ESCMAEAGR1',
       'TPNIVELINV', 'DTCADINF', 'MORTEPARTO', 
       'CAUSABAS_O','ano_obito', 'CID3']]
```
