```python
from pysus import SIA
import pandas as pd
import os
sia = SIA().load()
```

```python
variaveis = [
    "PA_UFMUN", "PA_CMP", "PA_PROC_ID", "PA_NIVCPL", "PA_OBITO", 
    "PA_ENCERR", "PA_PERMAN", "PA_ALTA", "PA_CIDPRI", "PA_CIDCAS", 
    "PA_IDADE", "PA_SEXO", "PA_RACACOR", "PA_MUNPCN", "PA_QTDPRO", 
    "PA_QTDAPR"
]
```

### Variáveis conforme documentação
- PA_CIDPRI: CID9 Principal (APAC ou BPA-I)
- PA_MUNPCN: Código da Unidade da Federação + Código do Município de residência do paciente ou do estabelecimento,
caso não se tenha a identificação do paciente, o que ocorre no (BPA)
- PA_QTDPRO: Quantidade Produzida (APRESENTADA)

### Variável QTD_PRODUZIDA = 90 para UF - BA para o município 110001

```python
files = sia.get_files("PA", uf="BA", year=2022, month=[5])
parquet = sia.download(files)
df = parquet.to_dataframe()
df = df[variaveis]
#df = df[df['PA_CIDPRI'].str.startswith('I', na=False)]
#df.query("PA_MUNPCN == '110001'")
```

```python
df['PA_QTDPRO'] = df['PA_QTDPRO'].astype(int)
```

```python
df.query("PA_MUNPCN == '292010' and PA_PROC_ID == '0401010015'")['PA_QTDPRO'].sum()
```

### Variável QTD_PRODUZIDA = 21 para UF - RO para o município 110001

```python
files = sia.get_files("PA", uf="RO", year=2022, month=[5])
parquet = sia.download(files)
df = parquet.to_dataframe()
df = df[variaveis]
df = df[df['PA_CIDPRI'].str.startswith('I', na=False)]
df.query("PA_MUNPCN == '110001'")
```
