```python
from pysus import SIA
import pandas as pd
import os
sia = SIA().load()
```

```python
variaveis = [
    "PA_UFMUN", "PA_CMP", "PA_PROC_ID", "PA_NIVCPL", "PA_OBITO", 
    "PA_ENCERR", "PA_PERMAN", "PA_ALTA", "PA_CIDPRI", "PA_CIDCAS", 
    "PA_IDADE", "PA_SEXO", "PA_RACACOR", "PA_MUNPCN", "PA_QTDPRO", 
    "PA_QTDAPR"
]
```

```python
sia = SIA().load()

ufs = [
    'AC', 'AL', 'AP', 'AM', 'BA', 'CE', 'DF', 'ES', 'GO',
    'MA', 'MT', 'MS', 'MG', 'PA', 'PB', 'PR', 'PE', 'PI',
    'RJ', 'RN', 'RS', 'RO', 'RR', 'SC', 'SP', 'SE', 'TO'
]

ufs = ['SP']
anos = [2022, 2023, 2024]

os.makedirs("saida_sia", exist_ok=True)

for uf in ufs:
    dados_uf = []

    for ano in anos:
        try:
            arquivos = sia.get_files(group="PA", year=ano, uf=uf)
            if not arquivos:
                print(f"Nenhum arquivo encontrado para {uf}/{ano}")
                continue

            for arq in arquivos:
                parquet = sia.download(arq)
                df = parquet.to_dataframe()
                df = df[variaveis]
                df = df[df['PA_CIDPRI'].str.startswith('I', na=False)]
                df["UF"] = uf
                df["ANO"] = ano
                dados_uf.append(df)
                print(f"Adicionado {uf}/{ano}")

        except Exception as e:
            print(f"Erro ao processar {uf}/{ano}: {e}")

    if dados_uf:
        df_uf = pd.concat(dados_uf, ignore_index=True)
        df_uf.to_csv(f"saida_sia/sia_PA_{uf}_2022_2024_.csv", index=False)
        print(f"Salvo: sia_PA_{uf}_2022_2024.csv")
```
