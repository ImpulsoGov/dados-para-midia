```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
import time

# Caminho para seu arquivo HTML
html_file_path = "/Users/juliasbardelatti/Desktop/macbook/julia/projetos/ImpulsaGov/Dados_midia/estudo_cancer_colo_utero/teste.html"

# Configurações do Chrome
options = Options()
#options.add_argument("--headless")  # Roda o navegador sem interface gráfica
options.add_argument("--window-size=1920,1080")  # Tamanho da janela
options.add_argument("--hide-scrollbars")  # Esconde barras de rolagem
options.add_argument("--force-device-scale-factor=2")  # Aumenta a densidade de pixels (tipo Retina)
options.add_argument("--disable-gpu")

# Inicializa o navegador
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# Abre o arquivo local (usando file://)
driver.get("file://" + html_file_path)

# Aguarda o carregamento do mapa
time.sleep(20)  # pode ajustar o tempo se necessário

# Tira o screenshot
driver.save_screenshot("mapa_screenshot.png")

# Encerra o navegador
driver.quit()
```
